Homework-2 Verilen listede bulunan string veri tipindeki öğeleri yeni_liste isimli listeye eklenir.

liste = ["Python",True,9,"3",8.4,"Hi-Kod","False",4.7]

```python
liste = ["Python", True, 9, "3", 8.4, "Hi-Kod", "False", 4.7]

yeni_liste = []

for i in liste:
    if type(i) == str:
        yeni_liste.append(i)

print(yeni_liste)
```

Homework-3 Enumerate methodunu araştırın ve aşağıdaki örneği enumerate methodu ile yapın.

 for index in range(len(meyveler)):


 print("{}. indexte bulunan meyve:{}".format(index,meyveler[index]))


**NOT**

burada len(meyveler) meyveler listesinin uzunluğunu verir,

range(len(meyveler)) liste elemanlarını indexler

for index in range(...) → index değerlerini tek tek döngüye sokar

Döngü şu şekilde işler:

1.tur → index = 0

2.tur → index = 1

3.tur → index = 2



```python
meyveler = ["elma", "armut", "çilek", "muz"]

for index, meyve in enumerate(meyveler):
    print("{}. indexte bulunan meyve: {}".format(index, meyve))
```

Burada enumerate methodu range(len komutlarını birlikte yapar diyebiliriz
