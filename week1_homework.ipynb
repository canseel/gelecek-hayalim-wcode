**Homework 2 - Operations**

İsimlerden oluşan üç değişkene yaş değerleri atanır. Belirlenen üç değişken birbiriyle karşılaştırma operatörleri ile karşılaştırılır. Bu karşılaştırmalara mantıksal operatörler de eklenir.

```python
ceyda = 27
mert = 44
hayri = 12 #isimlerden oluşan değişkenlere yaş değeri ataması yaptım
```

```python
# Karşılaştırma Operatörleri
if ceyda > hayri:
  print("Ceyda hayri'den büyüktür.")
else:
   print("Ceyda hayri'den küçüktür.")

if ceyda > mert:
  print("Ceyda mert'ten büyüktür.")
else:
   print("Ceyda mert'ten küçüktür.")

if hayri > mert:
  print("Hayri mert'ten büyüktür.")
else:
   print("Hayri mert'ten küçüktür.")


# Mantıksal Operatörler
if ceyda > hayri and ceyda > mert:
  print("Ceyda en yaşlıdır.")

if hayri > ceyda and hayri > mert:
  print("Hayri en yaşlıdır.")

if mert > hayri and mert > ceyda:
  print("Mert en yaşlıdır.")

# Alternatif olarak;

print("Ceyda 20'den büyük ve Hayri 25'ten küçük mü?", ceyda > 20 and hayri < 25)
print("Mert 35 yaşında mı ya da Hayri 20 yaşında mı?", mert == 35 or hayri == 20)
print("Ceyda 30 yaşında değil mi?", not ceyda == 30)



```

**Homework 5**

"Hi-Kod Veri Bilimi Atölyesi" ifadesini bir değişkene tanımlanır.

1.   İfadedeki her bir kelimeyi ("Hi-Kod", "Veri", "Bilimi", "Atölyesi") değişken içinden seçilir.
2.   İfadeyi hepsini büyük harf olacak hale çevrilir. ("HI-KOD VERİ BİLİMİ ATÖLYESİ")
3.  İfadeyi hepsini büyük harf olacak hale çevrilir.("hi-kod veri bilimi atölyesi")
"0123456789" ifadesindeki yalnızca çift sayıları ve yalnızca tek sayıları seçilir. ("02468", "13579")



```python
atolye = "Hi-Kod Veri Bilimi Atölyesi"
kelimeler = atolye.split() #split: ifadedeki kelimeleri ayırır
print("kelimeler:",kelimeler)
print("1.kelime:", kelimeler[0])
print("2.kelime:", kelimeler[1])
print("3.kelime:", kelimeler[2])
print("4.kelime:", kelimeler[3])  #her bir kelimeyi numaralandırdık diyebiliriz

buyuk_harf = atolye.upper()
print("Büyük harf:", buyuk_harf) #ifadeyi büyük harfe çevirme

kucuk_harf = atolye.lower()
print("Küçük harf:", kucuk_harf) #ifadeyi küçük harfe çevirme

#slicing

sayilar = "0123456789"
ciftler = sayilar[::2] #stringin başından başlayarak her 2. karakteri seç
tekler = sayilar[1::2] #1. karakterden başlayarak her 2. karakteri seç
print("Çift sayılar:", ciftler)
print("Tek sayılar:", tekler)

```
