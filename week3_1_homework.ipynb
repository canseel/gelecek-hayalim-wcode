Homework 1 - Kullanıcıdan pi değeri ve yarıçap bilgisi alarak dairenin alanını hesaplayan bir fonksiyon oluşturulur.

```python
def daire_alani_hesapla():
  pi = float(input("Lütfen pi değerini giriniz: "))
  yaricap = float(input("Lütfen yarıçap değerini giriniz: "))
  alan = pi * yaricap * yaricap
  print("Dairenin alanı:", alan)

#son olarak fonksiyonu çağırıyoruz
daire_alani_hesapla()
```

Homework 2 - Faktöriyel adında fonksiyon oluşturulur. Döngü kullanarak parametre olarak girilen sayının faktöriyeli hesaplanır. Format metodunu kullanılarak ekrana yazdırılır.

```python
def faktoriyel(n):
    baslangic = 1
    for i in range(1, n + 1):
        baslangic *= i
    print("{}. sayısının faktöriyeli = {}".format(n, baslangic))

faktoriyel(5)
```

Homework 4 - Doğum yılı ve isim bilgisi verilen fonksiyon kişinin emekli olup olmadığını söylesin.(Kişi 65 yaşında ise emekli olur.) Burada yaş hesabını yukarıdaki örnekteki fonksiyonu kullanarak yapsın.(Yani fonksiyon içinde fonksiyon kullanmanızı istiyorum :)) Kişi 65 yaşında ya da daha fazlaysa "Emekli oldunuz" yanıtını, 65 yaşından küçükse emekliliğine kaç yıl kaldığını da hesaplayarak "(isim) emekliliğine (yıl) kaldı." yanıtını versin.

```python
def yas_hesapla(dogum_yili):
    mevcut_yil = 2025
    return mevcut_yil - dogum_yili


def emeklilik_durumu(isim, dogum_yili):
    yas = yas_hesapla(dogum_yili)   # Bir önceki fonksiyonu burada kullandık

    if yas >= 65:
        print("Emekli oldunuz.")
    else:
        kalan = 65 - yas
        print("{} emekliliğine {} yıl kaldı.".format(isim, kalan))


# Fonksiyon çağrısı
emeklilik_durumu("Cansel", 2001)
```
