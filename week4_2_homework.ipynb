Sayılardan oluşan bir boyutlu array oluşturun. Arrayi oluştururken sayıların veri tipini integer olarak belirtelim.

Oluşturulan arrayin boyut, eleman sayısı bilgilerine bakalım.

Homework 2

Aşağıda verilen iki boyutlu ve üç boyutlu arrayi oluşturalım. Bu arraylerin boyut, eleman sayısı, satır, sütun bilgilerine ulaşalım.


İki boyutlu array:

[[1,2,6,7],

[4,3,9,5]]




Üç boyutlu array:

[[[7,5,14],[21,8,11]],

[[8,6,20],[14,3,9]]]

```python
import numpy as np

ikib = np.array([
    [1,2,6,7],[4,3,9,5]
    ])

print("Array:", ikib)
print("Boyut sayısı (ndim):", ikib.ndim)
print("Toplam eleman sayısı (size):", ikib.size)
print("Şekil (satır, sütun):", ikib.shape)
```

```python
import numpy as np

ucb = np.array([[[7,5,14],[21,8,11]],
                   [[8,6,20],[14,3,9]]])

print("Array:", ucb)
print("Boyut sayısı (ndim):", ucb.ndim)
print("Toplam eleman sayısı (size):", ucb.size)
print("Şekil (satır, sütun):", ucb.shape)
```

Homework 5

5 satır 3 sütunluk sıfırlardan ve birlerden oluşan iki tane iki boyutlu array oluşturalım. Bu arrayleri satır ve sütun bazında birleştirelim.

```python
import numpy as np

# Rastgele 0 ve 1'lerden oluşan 5x3 arrayler
arr1 = np.random.randint(0, 2, (5, 3))
arr2 = np.random.randint(0, 2, (5, 3))

print("Array 1:\n", arr1)
print("\nArray 2:\n", arr2)

#Satır bazında birleştirme
arr_vertical = np.vstack((arr1, arr2))
print("\nSatır bazında birleştirilmiş array (10x3):\n", arr_vertical)

#Sütun bazında birleştirme
arr_horizontal = np.hstack((arr1, arr2))
print("\nSütun bazında birleştirilmiş array (5x6):\n", arr_horizontal)
```
